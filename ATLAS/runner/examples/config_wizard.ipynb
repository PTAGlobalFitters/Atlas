# atlas-run config wizard

Pick a preset on **Start**, go through the tabs, then **Validate** and **Save**. Save writes the run config and its priors file into the same folder; run it with `atlas-run run <file>`.

The preview on the right is exactly what Save writes. Settings the menus do not cover can be added to the saved YAML by hand.

```
from ATLAS.runner.wizard import Wizard

wz = Wizard(save_path="/data/taylor_group/Nima/runs/my_run.yaml")
wz
```

Everything is also scriptable, e.g. to start from a preset, change one thing and save:

```
# wz.apply_preset("PTA GWB search (HD, gamma = 13/3)")
# wz.w["num_samples"].value = 10000
# wz.save("/data/taylor_group/Nima/runs/hd.yaml")
```
